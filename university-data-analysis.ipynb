#### Data Cleaning

```python
df['Annual Academic Cost (PKR)']
```

```python
# Clean and convert column to numeric
df['Annual Academic Cost (PKR)'] = pd.to_numeric(
    df['Annual Academic Cost (PKR)'].astype(str)
    .str.replace(',', '',regex=True),
    errors='coerce'
)

# Calculate median and fill missing values
median_cost = df['Annual Academic Cost (PKR)'].median()
df['Annual Academic Cost (PKR)'] = df['Annual Academic Cost (PKR)'].fillna(median_cost)
```

```python
df['Annual Academic Cost (PKR)']
```

```python
def extract_midpoint(rank):
    # Define a function named extract_midpoint that takes rank as input
    if pd.isna(rank) or rank == '':
        # Check if rank is missing (NaN) OR if rank is an empty string
        return None
        # Return None if there is no valid rank
    rank = str(rank).strip()
    # Convert rank to a string and remove extra spaces from the beginning/end
    if '-' in rank:
        # Check whether rank contains '-' (meaning it is a range like "351-400")
        parts = rank.split('-')
        # Split the range into two parts: "351" and "400"
        return int((int(parts[0]) + int(parts[1])) / 2)
        # Convert both parts to integers, add them, and divide by 2
        # Example: (351 + 400) / 2 = 375.5
    else:
        # If '-' is not present, rank is a single value like "114"
        return int(rank)
        # Convert the single rank from string to integer and return it
```

```python
df[' QS World University Ranking (2026) ']=df[' QS World University Ranking (2026) '].apply(extract_midpoint)
```

```python
df.columns=[
    'University','Location','Ranking','Type','Cost','Citations','Academic_Reputation','Research_Network','Employer_Reputation'
]

df.head(2)
```

```python
df.columns=df.columns.str.strip()
```

#### Key Findings & Conclusions

### Project Summary

This analysis examined the **Pakistan IT University Benchmarks Dataset**, comparing 15 universities across multiple performance metrics including academic costs, research output, and global rankings.

### Major Findings

1. **Cost Structure**
   - Academic costs vary significantly across universities, with private institutions generally charging much higher fees
   - There's a wide range from low-cost public universities to premium private institutions
   - The median cost balances out extreme outliers, providing a realistic mid-point reference

2. **Public vs Private Universities**
   - Pakistan has 10 public and 5 private universities in this dataset (67% public, 33% private)
   - Private universities charge significantly higher fees on average
   - Despite higher costs, private universities don't necessarily rank better globally

3. **Research Performance (Citations)**
   - Citation rates are relatively consistent across universities (80-88 citations per paper average)
   - **Key insight**: Higher cost does NOT guarantee more citations per paper
   - Research quality appears to be independent of tuition fees

4. **Global Rankings**
   - Top-ranked universities include NUST (rank 114) and COMSATS (rank 180)
   - Most universities cluster in the 351-400 ranking range
   - Better rankings correlate with academic reputation and research networks rather than cost alone

5. **Geographic Distribution**
   - Islamabad and Lahore dominate with majority of universities
   - These cities appear to be educational hubs in Pakistan

### Recommendations

- **For Students**: Higher tuition cost doesn't necessarily mean better research output or global rankings
- **For Universities**: Focus on research quality and international collaboration rather than premium pricing
- **For Policymakers**: Public universities demonstrate competitive research metrics, showing good ROI on public education investment

### Conclusion

The analysis reveals that university quality in Pakistan is multifaceted. While academic costs vary widely, the most important metrics for university excellence—research output, global reputation, and employer recognition—are not solely determined by tuition fees. This suggests that both public and private institutions have pathways to achieve international competitiveness.
